```python
#Q3.Connect to Azure SQL using your .env file. Print Connected successfully.


import pandas as pd
import numpy as np
import requests
import anthropic
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
import urllib
from dotenv import load_dotenv
import os
import requests
import pandas as pd
import json

load_dotenv()

conn = urllib.parse.quote_plus(
    f"DRIVER={{ODBC Driver 18 for SQL Server}};"
    f"SERVER={os.getenv('DB_SERVER')};"
    f"DATABASE={os.getenv('DB_NAME')};"
    f"UID={os.getenv('DB_USER')};"
    f"PWD={os.getenv('DB_PASSWORD')}"
)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={conn}")
```

```python
#4. Run a query returning all delivered orders from 2018. Print the row count.

df_revenue = pd.read_sql("""
    SELECT c.customer_state,
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(SUM(p.payment_value), 2) AS total_revenue
    FROM olist_orders_dataset o
    JOIN olist_customers_dataset c
    ON o.customer_id = c.customer_id
    JOIN olist_order_payments_dataset p
    ON o.order_id = p.order_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_state
    ORDER BY total_revenue DESC
""", engine)

print(f"Revenue data loaded: {len(df_revenue)} states")
print(df_revenue.head(10))
```

```python
#Q5. Run the top 10 most expensive items query. Store in a dataframe and print it.

top_10_expensive_items_df = pd.read_sql("""
    SELECT TOP 10 order_item_id, 
    price FROM olist_order_items_dataset
    ORDER BY price DESC 
""", engine)
print(top_10_expensive_items_df)
```
